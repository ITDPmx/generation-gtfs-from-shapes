## Codigo para analizar utas y paradas de  GTFS

```python
import pandas as pd
import geopandas as gpd
import geopandas as gpd
from shapely.geometry import LineString
```

```python
route = "7"
```

```python
path_gtfs = "./data-gtfs/"
```

### Routes

```python
path_routes = path_gtfs + "shapes.txt"
routes = pd.read_csv(path_routes)
```

```python
routes = routes[routes["shape_id"]== f"shape_{route}-1"]
routes.head()
```

```python
# 1. Ordenar los puntos por secuencia
routes = routes.sort_values(by="shape_pt_sequence")

# 2. Crear geometría LineString (usa lon, lat)
line = LineString(zip(routes["shape_pt_lon"], routes["shape_pt_lat"]))

# 3. Crear un GeoDataFrame con CRS WGS84 (EPSG:4326)
gdf = gpd.GeoDataFrame(
    {"shape_id": [routes["shape_id"].iloc[0]]},
    geometry=[line],
    crs="EPSG:4326"
)

# 4. Guardar a GeoJSON
gdf.to_file(f"route_line_{route}.geojson", driver="GeoJSON")
```

## Stops

```python
path_stops = path_gtfs + "stops.txt"
stops = pd.read_csv(path_stops)
```

```python
stops["route"] = stops["stop_id"].str.split("_").str[1]
```

```python
stops["sequence"] = stops["stop_id"].str.split("_").str[-1]
stops["sequence"] = stops["sequence"].astype(int)
```

```python
stops.head()
```

```python
stops = stops[stops["route"]== f"r{route}"]
stops.head()
```

```python
stops.to_csv(f"./stops_route_{route}.csv", index=False)
```
