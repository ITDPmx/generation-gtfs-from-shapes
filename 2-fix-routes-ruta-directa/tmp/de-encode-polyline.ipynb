```python
import polyline as pl
from shapely.geometry import LineString
import geopandas as gpd
```

```python


def encoded_polyline_to_linestring(enc: str, precision: int = 5) -> LineString:
    """
    Convierte un encoded polyline (Google) en una Shapely LineString.
    precision=5 es lo usual; usa 6 si tu polyline viene con 6 decimales.
    """
    # polyline.decode -> lista de (lat, lon)
    latlon = pl.decode(enc, precision=precision)
    # Shapely usa (x=lon, y=lat)
    lonlat = [(lon, lat) for lat, lon in latlon]
    return LineString(lonlat)

# --- Ejemplo de 1 polyline ---
encoded = "ihrfCn}ftQjA_Dl[vNk@|D}\\iOan@oXeRaI}JqDyFT}JzFyEhBmLfCwDtBeCvLaGhG_L`DcFvDyF~@uIk@qKSmEiBsL|@cDaBcNkD}MsDarBuj@a@sA}Ar@gEcDuBqEc@_Q?aPOuQM}Q_GoXwLhE{ImUWcAgFkLxKmGyg@ur@wZyb@Oi@?{@m@a@o@NSp@Tp@n@L`J|KlCrDfWr^}RhHgKdDIdAgATo@Yf@dAoH~BSlCEjIc@f@gDNYcHy@y@A{AkCLYsAgE^??fE_@XrAjCM@zAx@x@XbHfDOb@g@DkIRmCnH_Cg@eAn@XfAUHeAfKeD|RiHgWs^mCsDaJ}Ko@MUq@Rq@n@Ol@`@?z@Nh@vZxb@xg@tr@yKlGfFjLVbAzIlUvLiE~FnXL|QNtQ?`Pb@~PtBpEfEbD|As@`@rA`rBtj@|MrDbNjDbD`BrL}@lEhBpKRtIj@xF_AbFwD~KaD`GiGdCwLvDuBlLgCxEiB|J{FxFU|JpDdR`I`n@nX|\\hOj@}Dm[wN"
geom = encoded_polyline_to_linestring(encoded, precision=5)

gdf = gpd.GeoDataFrame(
    {"name": ["ruta_1"]},
    geometry=[geom],
    crs="EPSG:4326"
)

# Guardar a GeoJSON
out_path = "ruta_1.geojson"
gdf.to_file(out_path, driver="GeoJSON")
print(f"Guardado: {out_path}")
```
