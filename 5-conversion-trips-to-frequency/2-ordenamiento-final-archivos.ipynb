```python
import os
import shutil

def borrar_archivo(ruta):
    """
    Borra un archivo específico.
    
    ruta: path al archivo (ej: "./data/file.txt")
    """
    try:
        if os.path.isfile(ruta):
            os.remove(ruta)
            print(f"Archivo eliminado: {ruta}")
        else:
            print(f"No existe el archivo: {ruta}")
    except Exception as e:
        print(f"Error al borrar el archivo {ruta}: {e}")

        
def borrar_carpeta(carpeta):
    # Verifica si existe y bórrala
    if os.path.exists(carpeta):
        shutil.rmtree(carpeta)
        print(f"Carpeta eliminada: {carpeta}")
    else:
        print(f"La carpeta no existe: {carpeta}")

def crear_carpeta(ruta):
    """
    Crea una carpeta en la ruta especificada.
    Si la carpeta ya existe, no lanza error.
    """
    try:
        os.makedirs(ruta, exist_ok=True)
        print(f"Carpeta creada o ya existente: {ruta}")
    except Exception as e:
        print(f"Error al crear la carpeta: {e}")

def copiar_archivos(archivos, carpeta_destino):
    """
    Copia una lista de archivos a una carpeta destino.
    
    archivos: lista de rutas de archivos (ej: ["./data/a.txt", "./data/b.csv"])
    carpeta_destino: ruta de la carpeta destino (ej: "./backup/")
    """
    # Crear carpeta destino si no existe
    os.makedirs(carpeta_destino, exist_ok=True)

    for archivo in archivos:
        if os.path.isfile(archivo):
            shutil.copy(archivo, carpeta_destino)
            print(f"Copiado: {archivo} -> {carpeta_destino}")
        else:
            print(f"No existe o no es archivo: {archivo}")
```
