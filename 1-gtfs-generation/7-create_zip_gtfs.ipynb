# Create GTFS ZIP file

Compresses all GTFS .txt files into a single .zip archive.

```python
import json
from pathlib import Path as Path
import zipfile  
```

## Parameters

```python

# Path to params.json (same directory as this notebook)
_params_path = "../params.json"
with open(_params_path, encoding="utf-8") as f:
    p = json.load(f)

CITY = p["city"]
```

```python
# --- Carpeta GTFS ---
PATH_DIR_GTFS = Path(f"../data/{CITY}/gtfs-frequencies")
PATH_DIR_GTFS.mkdir(parents=True, exist_ok=True)
print(f"Salida: {PATH_DIR_GTFS.absolute()}")
```

## Compress GTFS files based on frequencies

```python

# Solo archivos .txt dentro de la carpeta GTFS frequencies
txt_files = sorted(PATH_DIR_GTFS.glob("*.txt"))
zip_path = PATH_DIR_GTFS.parent / "gtfs-frequencies.zip"

with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as zf:
    for f in txt_files:
        zf.write(f, f.name)

print(f"Zip creado: {zip_path}")
print(f"Archivos incluidos ({len(txt_files)}): {[f.name for f in txt_files]}")
```

```python
# Verificar contenido del zip
with zipfile.ZipFile(zip_path, "r") as zf:
    print("Contenido del zip:", zf.namelist())
```
