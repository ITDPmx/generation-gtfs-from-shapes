# Generate agency.txt and calendar.txt

Generates GTFS agency and calendar files from parameters.

```python
import json
from pathlib import Path 
import pandas as pd
```

## Parameters

```python
# Path to params.json (same directory as this notebook)
_params_path = "../params.json"
with open(_params_path, encoding="utf-8") as f:
    p = json.load(f)
```

```python
# --- Agency (operator) ---
CITY = p["city"]
AGENCY_NAME = p["agency"]["name"]
AGENCY_ID = p["agency"]["id"]
AGENCY_URL = p["agency"]["url"]
AGENCY_TIMEZONE = p["agency"]["timezone"]
AGENCY_LANG = p["agency"]["lang"]
```

```python
# --- Calendar: common validity and three services ---
START_DATE = p["calendar"]["start_date"]
END_DATE = p["calendar"]["end_date"]
SERVICE_ID = p["calendar"]["SERVICE_ID"]
# SERVICE_ID_PICO = p["calendar"]["service_id_pico"]
# SERVICE_ID_FINDE = p["calendar"]["service_id_finde"]
```

```python
# --- GTFS folder ---
PATH_DIR_GTFS = Path(f"../data/{CITY}/gtfs-frequencies")
PATH_DIR_GTFS.mkdir(parents=True, exist_ok=True)
print(f"Output: {PATH_DIR_GTFS.absolute()}")

# --- Processed folder ---
PATH_DIR_processed = Path(f"../data/{CITY}/processed")
PATH_DIR_processed.mkdir(parents=True, exist_ok=True)
print(f"Output: {PATH_DIR_processed.absolute()}")
```

## Generate agency.txt

```python
df_agency = pd.DataFrame([{
    "agency_id": AGENCY_ID,
    "agency_name": AGENCY_NAME,
    "agency_url": AGENCY_URL,
    "agency_timezone": AGENCY_TIMEZONE,
    "agency_lang": AGENCY_LANG,
}])

path_agency = PATH_DIR_GTFS / "agency.txt"
df_agency.to_csv(path_agency, index=False, encoding="utf-8")
print(f"Guardado: {path_agency}")
df_agency
```

## Generate calendar.txt (three calendars)

```python
# Mon-Fri = 1,1,1,1,1,0,0 
filas = [
    {
        "service_id": SERVICE_ID,
        "monday": 1, "tuesday": 1, "wednesday": 1, "thursday": 1, "friday": 1,
        "saturday": 0, "sunday": 0,
        "start_date": START_DATE, "end_date": END_DATE,
    }
]
```

```python
df_calendar = pd.DataFrame(filas)
path_calendar = PATH_DIR_GTFS / "calendar.txt"
df_calendar.to_csv(path_calendar, index=False, encoding="utf-8")
print(f"Guardado: {path_calendar}")
df_calendar
```

## Summary

```python
files = list(PATH_DIR_GTFS.glob("*.txt"))
print(f"Files in {PATH_DIR_GTFS}:")
for f in sorted(files):
    print(f"  - {f.name}")
```
