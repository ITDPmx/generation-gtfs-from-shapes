# Generación GTFS: agency.txt y calendar.txt

Notebook parametrizado para generar los archivos **agency** y **calendar** de un feed GTFS sintético, según la especificación en `files_gtfs_to_generate.md`. Los archivos se guardan en `data/gtfs-output`.

```python
import pandas as pd
from pathlib import Path
```

## Parámetros

```python
import json
from pathlib import Path as PathLib

# Ruta a params.json (mismo directorio que este notebook)
_params_path = PathLib.cwd() / "params.json"
if not _params_path.exists():
    _params_path = PathLib("params.json")
with open(_params_path, encoding="utf-8") as f:
    p = json.load(f)

# --- Agency (operadora) ---
CIUDAD = p["ciudad"]
AGENCY_NAME = p["agency"]["name"]
AGENCY_ID = p["agency"]["id"]
AGENCY_URL = p["agency"]["url"]
AGENCY_TIMEZONE = p["agency"]["timezone"]
AGENCY_LANG = p["agency"]["lang"]

# --- Calendar: vigencia común y tres servicios ---
START_DATE = p["calendar"]["start_date"]
END_DATE = p["calendar"]["end_date"]
SERVICE_ID_VALLE = p["calendar"]["service_id_valle"]
# SERVICE_ID_PICO = p["calendar"]["service_id_pico"]
# SERVICE_ID_FINDE = p["calendar"]["service_id_finde"]
```

```python
# Parámetros de calendar cargados desde params.json en la celda anterior
```

```python
# --- Carpeta GTFS ---
PATH_DIR_GTFS = Path(f"../data/{CIUDAD}/gtfs-output")
PATH_DIR_GTFS.mkdir(parents=True, exist_ok=True)
print(f"Salida: {PATH_DIR_GTFS.absolute()}")

# --- Carpeta proccesed ---
PATH_DIR_proccesed = Path(f"../data/{CIUDAD}/processed")
PATH_DIR_proccesed.mkdir(parents=True, exist_ok=True)
print(f"Salida: {PATH_DIR_proccesed.absolute()}")
```

## Generar agency.txt

```python
df_agency = pd.DataFrame([{
    "agency_id": AGENCY_ID,
    "agency_name": AGENCY_NAME,
    "agency_url": AGENCY_URL,
    "agency_timezone": AGENCY_TIMEZONE,
    "agency_lang": AGENCY_LANG,
}])

path_agency = PATH_DIR_GTFS / "agency.txt"
df_agency.to_csv(path_agency, index=False, encoding="utf-8")
print(f"Guardado: {path_agency}")
df_agency
```

## Generar calendar.txt (tres calendarios)

- **L_V_VALLE**: entre semana, hora valle  
- **L_V_PICO**: entre semana, hora de máxima demanda  
- **S_D**: sábado y domingo

```python
# Lun-Vie = 1,1,1,1,1,0,0  |  Sáb-Dom = 0,0,0,0,0,1,1
filas = [
    {
        "service_id": SERVICE_ID_VALLE,
        "monday": 1, "tuesday": 1, "wednesday": 1, "thursday": 1, "friday": 1,
        "saturday": 0, "sunday": 0,
        "start_date": START_DATE, "end_date": END_DATE,
    }
]
```

```python
df_calendar = pd.DataFrame(filas)
path_calendar = PATH_DIR_GTFS / "calendar.txt"
df_calendar.to_csv(path_calendar, index=False, encoding="utf-8")
print(f"Guardado: {path_calendar}")
df_calendar
```

## Resumen

```python
archivos = list(PATH_DIR_GTFS.glob("*.txt"))
print(f"Archivos en {PATH_DIR_GTFS}:")
for f in sorted(archivos):
    print(f"  - {f.name}")
```
