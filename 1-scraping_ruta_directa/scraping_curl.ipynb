```python
import requests

def obtener_routes_names(slug_ciudad):
    """
    Realiza una petición a la API de buz.yt para obtener información
    de una ciudad específica mediante su slug.
    """
    url = "https://api.buz.yt/graphql/"

    headers = {
        "Referer": "https://rutadirecta.com/",
        "User-Agent": "Mozilla/5.0 (iPhone; CPU iPhone OS 18_5 like Mac OS X) AppleWebKit/605.1.15 (KHTML, like Gecko) Version/18.5 Mobile/15E148 Safari/604.1",
        "Accept": "application/json, text/plain, */*",
        "Content-Type": "application/json",
        "access_token": "" 
    }

    payload = {
        "operationName": "getCityBySlug",
        "variables": {"slug": slug_ciudad},
        "query": """query getCityBySlug($slug: String) {
          city(slug: $slug) {
            name
            lat
            lng
            slug
            state {
              iso2
              name
              __typename
            }
            entity {
              id
              __typename
            }
            routes(isActive: true) {
              id
              shortName
              longName
              slug
              color
              type
              __typename
            }
            __typename
          }
        }"""
    }

    try:
        response = requests.post(url, headers=headers, json=payload)
        # Lanza una excepción si la respuesta es un error (4xx o 5xx)
        response.raise_for_status() 
        return response.json()
    except requests.exceptions.RequestException as e:
        print(f"Error al realizar la petición: {e}")
        return None

```

```python
import pandas as pd
import time

# 1. Lista para ir acumulando los datos de las rutas
todas_las_rutas = []

print(f"Iniciando descarga de rutas para {len(df_nombres_ciudades)} ciudades...")

# 2. Iterar sobre cada fila del DataFrame original
for index, fila in df_nombres_ciudades.iterrows():
    slug = fila['slug']
    print(f"Consultando: {slug}...", end="\r")
    
    # Llamamos a tu función
    resultado = obtener_routes_names(slug)
    
    # Validamos que la respuesta tenga datos
    if resultado and "data" in resultado and resultado["data"]["city"]:
        datos_ciudad = resultado["data"]["city"]
        nombre_ciudad = datos_ciudad["name"]
        rutas = datos_ciudad["routes"]
        
        # 3. Iterar sobre las rutas de esta ciudad específica
        for ruta in rutas:
            # Añadimos información de la ciudad a cada ruta para saber a dónde pertenece
            ruta['ciudad_nombre'] = nombre_ciudad
            ruta['ciudad_slug'] = slug
            todas_las_rutas.append(ruta)
            
    # Opcional: una pequeña pausa para no saturar la API
    time.sleep(0.5)

# 4. Generar el nuevo DataFrame con todas las rutas recolectadas
df_rutas_final = pd.DataFrame(todas_las_rutas)

# Reorganizar columnas para que la ciudad aparezca al principio
columnas = ['ciudad_nombre', 'ciudad_slug'] + [c for c in df_rutas_final.columns if c not in ['ciudad_nombre', 'ciudad_slug']]
df_rutas_final = df_rutas_final[columnas]

print(f"\nProceso completado. Se encontraron {len(df_rutas_final)} rutas en total.")

# Mostrar los primeros resultados
df_rutas_final.head()
```

## Get routes shapes by routes

```python
import pandas as pd
import requests
import time
import polyline
from shapely.geometry import LineString
```

```python


# --- 1. Definición de funciones (Tus funciones proporcionadas) ---

def get_route_by_id(bus_id, access_token=""):
    url = "https://api.buz.yt/graphql/"
    headers = {
        "Referer": "https://rutadirecta.com/",
        "User-Agent": "Mozilla/5.0 (iPhone; CPU iPhone OS 18_5 like Mac OS X) AppleWebKit/605.1.15 (KHTML, like Gecko) Version/18.5 Mobile/15E148 Safari/604.1",
        "Accept": "application/json, text/plain, */*",
        "Content-Type": "application/json",
    }
    if access_token:
        headers["access_token"] = access_token
    
    payload = {
        "operationName": "getRouteById",
        "variables": {"id": bus_id},
        "query": """query getRouteById($id: ID!) {
          route(id: $id) {
            id
            encodedLine
            shortName
            longName
            slug
            type
            color
            trips {
              headsign
              stops {
                name
                lat
                lng
              }
            }
          }
        }"""
    }
    
    try:
        resp = requests.post(url, headers=headers, json=payload)
        resp.raise_for_status()
        data = resp.json()
        return data["data"]["route"]
    except Exception as e:
        print(f"Error en ID {bus_id}: {e}")
        return None

def safe_decode(encoded):
    if not encoded or not isinstance(encoded, str):
        return None
    # Limpieza de escapes dobles comunes en JSON strings
    s_clean = encoded.replace("\\\\", "\\")
    try:
        coords = polyline.decode(s_clean)
        # Convertimos a (Longitud, Latitud) para formato GeoJSON/Shapely
        coords_lonlat = [(lon, lat) for lat, lon in coords]
        return LineString(coords_lonlat) if coords_lonlat else None
    except Exception:
        return None
```

```python
import pandas as pd
import requests
import time
from concurrent.futures import ThreadPoolExecutor, as_completed
from tqdm import tqdm
```

```python
# --- 1. Función trabajadora ---
def procesar_ruta(args):
    """
    Función que será ejecutada por cada hilo.
    Recibe una tupla (route_id, nombre_ciudad)
    """
    route_id, nombre_ciudad = args
    try:
        # Llamamos a tu función original (asegúrate de que esté definida arriba)
        detalle = get_route_by_id(bus_id=route_id, access_token="")
        
        if detalle:
            detalle['ciudad_nombre'] = nombre_ciudad
            return detalle
    except Exception as e:
        # Silenciamos el error para no romper la barra de progreso, 
        # pero podrías loguearlo si es necesario
        return None
    return None
```

## Descarga de detalles por ruta

```python
# --- 2. Configuración del procesamiento en paralelo ---

# Definimos cuántas peticiones simultáneas queremos (recomiendo entre 5 y 10)
# Un número muy alto podría causar que la API te bloquee temporalmente (Error 429)
MAX_WORKERS = 8 

data_by_bus_list = []
# Preparamos los argumentos como una lista de tuplas [(id1, ciudad1), (id2, ciudad2), ...]
tareas = list(zip(df_rutas_final["id"], df_rutas_final["ciudad_nombre"]))

print(f"Iniciando descarga paralela ({MAX_WORKERS} hilos) para {len(tareas)} rutas...")

# Usamos ThreadPoolExecutor para manejar los hilos
with ThreadPoolExecutor(max_workers=MAX_WORKERS) as executor:
    # Mapeamos las tareas y usamos tqdm para ver el progreso
    futures = [executor.submit(procesar_ruta, tarea) for tarea in tareas]
    
    for future in tqdm(as_completed(futures), total=len(tareas), desc="Descargando"):
        resultado = future.result()
        if resultado:
            data_by_bus_list.append(resultado)

# --- 3. Normalización y Limpieza final ---
if data_by_bus_list:
    df_detalles = pd.json_normalize(data_by_bus_list)

    # Reordenar columnas
    if 'ciudad_nombre' in df_detalles.columns:
        cols = ['ciudad_nombre'] + [c for c in df_detalles.columns if c != 'ciudad_nombre']
        df_detalles = df_detalles[cols]
    
    df_detalles = df_detalles.drop(columns=["entity", "__typename"], errors="ignore")

    print(f"\nProceso finalizado. Se obtuvieron {len(df_detalles)} rutas.")
else:
    print("\nNo se recuperaron datos.")

df_detalles.head(20)
```

```python
gdf_route.drop(columns=["id"], inplace=True)
gdf_route.rename(columns={
                            "shortName": "data.route.shortName",
                            "longName": "data.route.longName",                            
                            }, inplace=True)
```

```python
gdf_route["data.route.shortName"] = gdf_route["data.route.shortName"].fillna(
    gdf_route["data.route.longName"]
)


gdf_route["data.route.longName"] = gdf_route["data.route.longName"].fillna(
    gdf_route["data.route.shortName"]
)
```

```python
new_name_long = gdf_route["data.route.shortName"] + "_" + gdf_route["data.route.longName"]

gdf_route["data.route.shortName"] = gdf_route["data.route.longName"]

gdf_route["data.route.longName"] = new_name_long
```

```python
import os
import geopandas as gpd

output_folder = "./data/proc/"
# 2. Obtener la lista de ciudades únicas
ciudades = gdf_route['ciudad_nombre'].unique()

print(f"Iniciando exportación de {len(ciudades)} ciudades...")

for ciudad in ciudades:
    # Filtrar el GeoDataFrame para la ciudad actual
    gdf_ciudad = gdf_route[gdf_route['ciudad_nombre'] == ciudad]
    
    # Limpiar el nombre de la ciudad para el nombre del archivo
    # (Quitamos espacios y caracteres especiales básicos)
    nombre_archivo = ciudad.replace(" ", "_").lower()
    filepath = os.path.join(output_folder, f"{nombre_archivo}.geojson")
    
    try:
        # Guardar como GeoJSON
        # El driver 'GeoJSON' es el estándar para estos archivos
        gdf_ciudad.to_file(filepath, driver='GeoJSON')
        print(f"✅ Guardado: {filepath} ({len(gdf_ciudad)} rutas)")
    except Exception as e:
        print(f"❌ Error al guardar {ciudad}: {e}")

print("\n¡Exportación completada!")
```
